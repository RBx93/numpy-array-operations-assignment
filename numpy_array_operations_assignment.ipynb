Problem 1

In this part of the assignment, Numpy arrays. Create a 1D NumPy array of
shape [1x5] with random values drawn from a uniform distribution. Perform
the following operations:
• Compute the mean and standard deviation of the array.
• Reshape the array into a 2D array with 5 rows and 1 column.
• Add 5 to each element in the array and print the result.
• Compute the dot product of this reshaped array with itself.
After each step, print the resulting value.

```python
import numpy as numPyArray

print("Problem 1: \n")

#create a 1D numPy array of shape [1X5] with random values
oneDArray = numPyArray.random.rand(5)
print("Results for a 1D NumPy array of shape [1x5] with random values drawn from a uniform distribution: ",oneDArray, "\n")

#Compute mean
meanOneDArray = numPyArray.mean(oneDArray)
print("Mean: ", meanOneDArray, "\n")

#Compute Standard Deviation
stdDevOneDArray = numPyArray.std(oneDArray)
print("Standard Deviation: ", stdDevOneDArray, "\n")

#Reshape the array into a 2D array with 5 rows and 1 column
twoDArray = oneDArray.reshape(5,1)
print("Reshape the array into a 2D array with 5 rows and 1 column: \n", twoDArray, "\n")

#Add 5 to each element in the array and print the result
addValueTwoDArray = twoDArray + 5
print("Add 5 to each element in the array and print the result:\n", addValueTwoDArray, "\n")

#Compute the dot product of 2D array
dotProductTwoDArray = numPyArray.dot(addValueTwoDArray.T, addValueTwoDArray)
print("Output for Dot Product: ", dotProductTwoDArray, "\n")





```

Problem 2

In this part of the assignment, you will be working with visualizing plots with
Matplotlib. Generate a set of x values ranging from 0 to 100 with an
increment of 0.1 using NumPy. Compute the corresponding y values using
the function y=sin(x). Plot the sine wave using Matplotlib, and add
appropriate labels for the x-axis, y-axis, and a title for the plot. Save the plot
as a PNG file named “sine_wave.png”

```python
import numpy as numPyArray
import matplotlib.pyplot as plotGraphTool
from google.colab import files

print("Problem 2: \n")
#Generate a set of x values ranging from 0 to 100 with an increment of 0.1 using NumPy.
xValue = numPyArray.arange(0,100, 0.1)
print("Generated a set of x values ranging from 0 to 100 with an increment of 0.1 using NumPy: \n", xValue, "\n")

#Compute the corresponding y values using the function y=sin(x).
yValue = numPyArray.sin(xValue)
print("Compute the y values using y = sin(x): \n", yValue, "\n")

#Plot the sine wave using Matplotlib, and add appropriate labels for the x-axis, y-axis, and a title for the plot.
plotGraphTool.plot(xValue, yValue, label="y = sin(x)", color="green")

plotGraphTool.xlabel("x values") #Plot x
plotGraphTool.ylabel("y values") #Plot y
plotGraphTool.title("Sin(x) Wave") #Label Title bar
plotGraphTool.grid(True) #show grid
plotGraphTool.legend()  #show legend

plotGraphTool.savefig("sine_wave.png")
plotGraphTool.show()
files.download("sine_wave.png") #export file
```

Problem 3

In this part of the assignment, you will be integrating NumPy and Matplotlib
to analyze and visualize data. Perform the following steps:
• Create two NumPy arrays: one for x values ranging from 0 to 100 with
an increment of 1, and another for y values that represent a quadratic
function y=0.5x2 +2x+1.
• Plot the quadratic function using Matplotlib with appropriate labels and
a legend.
• Add gridlines to the plot and display it with a line style of your choice.
• Save the plot as a PDF file named quadratic_function.pdf.


```python
import numpy as numPyArray
import matplotlib.pyplot as plotGraphTool
from google.colab import files

print("Problem 3: \n")
# Step 1: Create x values from 0 to 100 with increment of 1
xValue = numPyArray.arange(0, 101, 1)

# Step 2: Calculate y = 0.5x^2 + 2x + 1
yValue = 0.5 * xValue**2 + 2 * xValue + 1

# Step 3:
plotGraphTool.figure(figsize=(10, 6))
plotGraphTool.plot(xValue, yValue, linestyle='--', color="green", label="y = 0.5x² + 2x + 1") #Plot the quadratic function
plotGraphTool.xlabel("x values")
plotGraphTool.ylabel("y values")
plotGraphTool.title('Quadratic Function')
plotGraphTool.grid(True)
plotGraphTool.legend()

plotGraphTool.savefig('quadratic_function.pdf') #Save the plot as a PDF

files.download('quadratic_function.pdf') #Download the PDF
```
